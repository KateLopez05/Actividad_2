#### CONSULTAS SELECT Y GROPU BY
Miremoslo de esta manera el select en las consultas son el "que" quiero ver dentro del gran registro de datos, el Group es el "como" queremos ver esos datos y el order by es para ordenar los datos.

```python
%sql
SELECT 
    d.gravedad,
    COUNT(f.radicado) as total_accidentes
FROM catalog_movilidad.schema_transito.tabla_incidente f
JOIN catalog_movilidad.schema_transito.tabla_incidente d 
    ON f.OBJECTID = d.OBJECTID
GROUP BY d.gravedad
ORDER BY total_accidentes DESC;
```
